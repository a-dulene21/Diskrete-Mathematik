1. Was bedeutet die Formel?
- \(P(B\mid A)\): Wahrscheinlichkeit von B, unter der Bedingung, dass A bereits eingetreten ist.
- \(P(A\cap B)\): Wahrscheinlichkeit, dass A und B gleichzeitig eintreten.
- \(P(A)\): Wahrscheinlichkeit, dass A eintritt.
