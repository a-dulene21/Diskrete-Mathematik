P(B=1)=0.20 + 0.20*R

- Wenn R = 0 ist, setzen wir 0 ein.
- Wenn R = 1 ist, setzen wir 1 ein.


```python


B=0.20 +0.40*R
print("P(B=1 | R=0) ", (0.20 +0.40*0))
print("P(B=1 |R=1)",  (0.20 +0.40*1) )

R=0
A=0.10 + 0.15*R + 0.25*B
print("P(A=1 |R=0) = ", A )

R=1
A=0.10 + 0.15*R + 0.25*B
print("P(A=1 |R=1) = ", A )

H=0.04 + 0.40*A
print("P(H=1) = ", H)
```
