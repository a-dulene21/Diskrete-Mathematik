```python
import numpy as np
```

```python
rng = np.random.default_rng(1); 
n = 100_000
R = rng.random(n) < 0.30
B = rng.random(n) < 0.20 + 0.40*R
A = rng.random(n) < 0.10 + 0.15*R + 0.25*B
H = rng.random(n) < 0.04 + 0.40*A
```

**1. Variable R**

```python
# Berechnung der bedingten Wahrscheinlichkeiten
p_H_given_R1 = np.mean(H[R == 1])
p_H_given_R0 = np.mean(H[R == 0])

print(f"P(H | R = 1) = {p_H_given_R1:.4f}")
print(f"P(H | R = 0) = {p_H_given_R0:.4f}")
```

#### 2. Prüfen Sie H ⫫ R | A: P(H | R, A) für A = 0 und A = 1
