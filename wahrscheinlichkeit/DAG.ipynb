#### Relation → Graph → DAG: Wie stark und wie zufällig?

Jeder Knoten wird zu einer Zufallsvariablen. Jede Kante beschreibt eine statistische Abhängigkeit oder einen möglichen Einfluss.

**1. Relation: Was hängt miteinander zusammen?**

**Eine Relation** beschreibt, welche Knoten miteinander verbunden sind.

**2. Graph: Wie stark ist die Beziehung?**

Jetzt bekommt jeder Knoten eine Zufallsvariable.
Eine Kante kann beispielsweise eine statistische Abhängigkeit ausdrücken.
