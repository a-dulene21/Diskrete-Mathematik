### Diabetes Daten bearbeiten

**nx.DiGraph** bedeutet, dass es sich um einen gerichteten Graphen handelt.



```python
import networkx as nx
import numpy as np

H = nx.DiGraph([
    ("R", "B"),
    ("R", "A"),
    ("B", "A"),
    ("A", "H")
])

K = nx.DiGraph([
    (2, 1),
    (2, 3),
    (3, 2),
    (5, 1),
    (5, 4),
    (6, 2),
    (6, 3),
    (6, 5)
])
```

##1. **\(A^2\) für den Knauer-Graphen** – wo steht die 2?

Die 2 steht als Exponent. Sie bedeutet, dass du die Adjazenzmatrix \(A\) mit sich selbst multiplizierst.

- \(A[i,j]=1\): Es gibt eine Kante von Knoten \(i\) nach Knoten \(j\).
- \(A^2[i,j]\): Anzahl der gerichteten Wege der Länge 2 von \(i\) nach \(j\).
Wichtig: Die Reihenfolge der Knoten in der Matrix ist 1, 2, 3, 4, 5, 6.

1. A² für den Knauer-Graphen – wo steht die 2?
Hier geht es um die Adjazenzmatrix und ihre zweite Potenz.

Die Adjazenzmatrix beschreibt, welche Knoten direkt miteinander verbunden sind.
- 1 bedeutet: Es gibt eine Kante.
- 0 bedeutet: Es gibt keine Kante.
Die 2 in \(A^2\) steht als Hochzahl. Sie bedeutet, dass die Matrix mit sich selbst multipliziert wird:

```python
A = nx.to_numpy_array(K, nodelist=sorted(K.nodes()), dtype=int)

A2 = np.linalg.matrix_power(A, 2)

print("Adjazenzmatrix A:")
print(A)

print("A²:")
print(A2)
```

Wie liest du diese Matrix?

Die Zeilen zeigen den Startknoten, die Spalten den Zielknoten.

Zum Beispiel ist der Eintrag in Zeile 2, Spalte 1 gleich 1. Das bedeutet: Es gibt den Pfeil \(2 \to 1\).

Der Eintrag in Zeile 1, Spalte 2 ist 0. Das bedeutet: Es gibt keinen Pfeil \(1 \to 2\).

```python
A = nx.to_numpy_array(
    K,
    nodelist=sorted(K.nodes()),
    dtype=int
)

print(A)
```

2. **Transitive Hülle mit \(W^+\)** aus V1 vergleichen

```python
TC = nx.transitive_closure(H)

print("Kanten der transitiven Hülle:")
print(sorted(TC.edges()))
```

### 3. Ist H oder K azyklisch? Finde einen Kreis in K.

H ist azyklisch: Es gibt keinen gerichteten Kreis.

K enthält den Kreis \(2 \to 3 \to 2\).

```python
print("H azyklisch:", nx.is_directed_acyclic_graph(H))
print("K azyklisch:", nx.is_directed_acyclic_graph(K))

print("Kreis in K:", nx.find_cycle(K))
```
