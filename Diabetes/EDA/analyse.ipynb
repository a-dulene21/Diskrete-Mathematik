# Recherche: Diabetes 130-US Hospitals (1999–2008)

## 1. Herkunft der Daten

Der Datensatz stammt aus 130 Krankenhäusern und integrierten Krankenhausnetzwerken in den USA. Er umfasst medizinische Behandlungsdaten aus den Jahren 1999 bis 2008. Die Daten wurden aus elektronischen Patientenakten (Electronic Health Records) gewonnen und 2014 im UCI Machine Learning Repository veröffentlicht.

Der Datensatz enthält 101.766 Krankenhausaufenthalte und 47 Variablen.

## 2. Echte oder simulierte Daten?

Es handelt sich um echte historische Behandlungsdaten und nicht um simulierte Daten. Die Patientendaten wurden anonymisiert. Trotzdem enthält der Datensatz sensible Informationen wie Alter, Geschlecht und ethnische Zugehörigkeit.

## 3. Datenqualität

Der Datensatz bietet viele Informationen über Patienten, Diagnosen, Medikamente, Laborwerte und Krankenhausaufenthalte.

Es gibt jedoch einige Probleme:
- Fehlende Werte, insbesondere bei Gewicht, Versicherungsangaben und medizinischer Fachrichtung.
- Unterschiedlich große Patientengruppen.
- Mehrere Krankenhausaufenthalte können zur gleichen Person gehören.
- Die Daten sind relativ alt und stammen aus den Jahren 1999 bis 2008.
- Die Daten bilden nicht alle Menschen mit Diabetes ab, sondern nur bestimmte Krankenhausaufenthalte.

Deshalb müssen die Daten vor der statistischen Auswertung geprüft und bereinigt werden.

## 4. Eignung für statistische Auswertungen

Mit 101.766 Krankenhausaufenthalten ist die Datenmenge grundsätzlich ausreichend für statistische Analysen. Möglich sind beispielsweise Häufigkeitsanalysen, Gruppenvergleiche, Chi-Quadrat-Tests und logistische Regressionen.

Die Ergebnisse müssen jedoch im Zusammenhang mit der Auswahl der Patienten und der Datenqualität interpretiert werden.

## 5. Wissenschaftliche Literatur

Eine wichtige Veröffentlichung ist:

Strack, B. et al. (2014): *Impact of HbA1c Measurement on Hospital Readmission Rates: Analysis of 70,000 Clinical Database Patient Records.* BioMed Research International. DOI: 10.1155/2014/781670.

Die Studie untersucht den Zusammenhang zwischen der Messung des HbA1c-Wertes und der Wiederaufnahme von Diabetespatienten ins Krankenhaus.

Eine weitere wichtige Veröffentlichung ist:

*The 30-days hospital readmission risk in diabetic patients: predictive modeling with machine learning classifiers* (2021).

Diese Studie untersucht die Vorhersage von Krankenhaus-Wiederaufnahmen mithilfe von Machine Learning.
